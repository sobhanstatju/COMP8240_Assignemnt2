```python
import numpy as np
import pandas as pd

N = 10000
rng = np.random.default_rng(42)

new_data = pd.DataFrame()

# 2. Child sex
new_data["child_sex"] = rng.choice(
    ["Male", "Female"],
    size=N, p=[0.503, 0.497]
)

# 3. Child age (months)
new_data["child_age"] = rng.choice(
    ["24-35", "36-47", "48-59"],
    size=N, p=[0.366, 0.355, 0.279]
)

# 4. Birth order
new_data["birth_order"] = rng.choice(
    ["1", "2-3", "4-6", ">=7"],
    size=N, p=np.array([0.240, 0.401, 0.268, 0.092]) / 1.001
)

# 5. Preceding birth interval
new_data["preceding_birth_interval"] = rng.choice(
    ["No preceding birth", "<24", "≥24"],
    size=N, p=np.array([0.240, 0.122, 0.637]) / 0.999
)

# 6. Baby postnatal check
new_data["baby_postnatal_check"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.236, 0.764]
)

# 7. Birth weight
new_data["birth_wt"] = rng.choice(
    ["Low", "Normal", "High (Macrosomia)", "Unknown"],
    size=N, p=[0.067, 0.636, 0.068, 0.229]
)

# 8. Vitamin A supplementation
new_data["supplemented_VitA"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.538, 0.462]
)

# 9. Intestinal parasite treatment
new_data["drug_intestinal_parasite"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.549, 0.451]
)

# 10. Early childhood disease
new_data["EaChDisease"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.268, 0.732]
)

# 11. Cough
new_data["cough"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.167, 0.833]
)

# 12. Diarrhea
new_data["diarrhea"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.078, 0.922]
)

# 13. Fever
new_data["fever"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.125, 0.875]
)

# 14. Stunting
new_data["stunted"] = rng.choice(
    ["Normal", "Stunted"],
    size=N, p=[0.742, 0.258]
)

# 15. Wasting
new_data["wasted"] = rng.choice(
    ["Normal", "Wasted"],
    size=N, p=[0.961, 0.039]
)

# 16. Underweight
new_data["underwt"] = rng.choice(
    ["Normal", "Underweight"],
    size=N, p=[0.879, 0.121]
)

# 17. Overweight
new_data["overwt"] = rng.choice(
    ["Normal", "Overweight"],
    size=N, p=[0.982, 0.018]
)

# 18. Mother's current age
new_data["current_age"] = rng.choice(
    ["<20", "20-35", "36-49"],
    size=N, p=[0.027, 0.743, 0.230]
)

# 19. Mother's education
new_data["education_level"] = rng.choice(
    ["No education", "Primary", "Secondary or above"],
    size=N, p=np.array([0.195, 0.487, 0.317]) / 0.999
)

# 20. Father's education
new_data["fathers_edu"] = rng.choice(
    ["No education", "Primary", "Secondary or above"],
    size=N, p=[0.175, 0.484, 0.341]
)

# 21. Marital status
new_data["marital_status"] = rng.choice(
    ["Married", "Single"],
    size=N, p=[0.596, 0.404]
)

# 22. Antenatal care visits
new_data["number_ANC"] = rng.choice(
    ["No ANC visit", "1-3", ">=4"],
    size=N, p=[0.102, 0.264, 0.634]
)

# 23. Working status
new_data["working_status"] = rng.choice(
    ["Working", "Not working"],
    size=N, p=[0.543, 0.457]
)

# 24. Wealth index
new_data["wealth_Index"] = rng.choice(
    ["Poor", "Medium", "Rich"],
    size=N, p=[0.431, 0.182, 0.387]
)

# 25. Number of household members
new_data["No_HH_Member"] = rng.choice(
    ["2-4", "5-7", ">7"],
    size=N, p=np.array([0.333, 0.457, 0.209]) / 0.999
)

# 26. Distance to health facility
new_data["distance_to_facility"] = rng.choice(
    ["Big problem", "Not a big problem"],
    size=N, p=[0.333, 0.667]
)

# 27. Health insurance
new_data["health_insurance"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.038, 0.962]
)

# 28. Water source
new_data["water_source"] = rng.choice(
    ["Improved", "Unimproved"],
    size=N, p=[0.675, 0.325]
)

# 29. Sanitation facility
new_data["sanitation_facility"] = rng.choice(
    ["Improved", "Unimproved"],
    size=N, p=[0.583, 0.417]
)

# 30. Household head sex
new_data["HH_head_sex"] = rng.choice(
    ["Male", "Female"],
    size=N, p=[0.749, 0.251]
)

# 31. Media exposure
new_data["media_exposure"] = rng.choice(
    ["Exposed", "Not exposed"],
    size=N, p=[0.674, 0.326]
)

# 32. Internet use
new_data["internet_use"] = rng.choice(
    ["Yes", "No"],
    size=N, p=[0.233, 0.767]
)

# 33. Residence
new_data["residence"] = rng.choice(
    ["Urban", "Rural"],
    size=N, p=[0.311, 0.689]
)






```

```python
import numpy as np
import pandas as pd
from scipy.special import expit
from scipy.optimize import brentq

# =====================================================
# 1. Feature importance from the chart (approximate)
# =====================================================

feature_importance = {
    "child_sex": 0.0226,
    "child_age": 0.0434,
    "birth_order": 0.0386,
    "preceding_birth_interval": 0.0304,
    "EaChDisease": 0.0126,
    "cough": 0.0177,
    "diarrhea": 0.0106,
    "current_age": 0.0238,
    "education_level": 0.0490,
    "marital_status": 0.0198,
    "working_status": 0.0240,
    "wealth_Index": 0.0352,
    "No_HH_Member": 0.0404,
    "distance_to_facility": 0.0275,
    "sanitation_facility": 0.0251,
    "HH_head_sex": 0.0238,
    "media_exposure": 0.0250,
    "internet_use": 0.0489,
    "residence": 0.0313
}

# =====================================================
# 2. Encode predictors for the simulation
# =====================================================

# These scores are assumptions, not estimated DHS effects.

effects = pd.DataFrame(index=new_data.index)

effects["child_sex"] = new_data["child_sex"].map({
    "Male": 0,
    "Female": 1
})

effects["child_age"] = new_data["child_age"].map({
    "24-35": 0,
    "36-47": 0.5,
    "48-59": 1
})

effects["birth_order"] = new_data["birth_order"].map({
    "1": 1,
    "2-3": 0.7,
    "4-6": 0.4,
    ">=7": 0
})

effects["preceding_birth_interval"] = (
    new_data["preceding_birth_interval"].map({
        "No preceding birth": 0.5,
        "<24": 0,
        "≥24": 1
    })
)

effects["EaChDisease"] = new_data["EaChDisease"].map({
    "Yes": 0,
    "No": 1
})

effects["cough"] = new_data["cough"].map({
    "Yes": 0,
    "No": 1
})

effects["diarrhea"] = new_data["diarrhea"].map({
    "Yes": 0,
    "No": 1
})

effects["current_age"] = new_data["current_age"].map({
    "<20": 0,
    "20-35": 1,
    "36-49": 0.5
})

effects["education_level"] = new_data["education_level"].map({
    "No education": 0,
    "Primary": 0.5,
    "Secondary or above": 1
})

effects["marital_status"] = new_data["marital_status"].map({
    "Married": 1,
    "Single": 0
})

effects["working_status"] = new_data["working_status"].map({
    "Working": 1,
    "Not working": 0
})

effects["wealth_Index"] = new_data["wealth_Index"].map({
    "Poor": 0,
    "Medium": 0.5,
    "Rich": 1
})

effects["No_HH_Member"] = new_data["No_HH_Member"].map({
    "2-4": 1,
    "5-7": 0.5,
    ">7": 0
})

effects["distance_to_facility"] = (
    new_data["distance_to_facility"].map({
        "Big problem": 0,
        "Not a big problem": 1
    })
)

effects["sanitation_facility"] = (
    new_data["sanitation_facility"].map({
        "Improved": 1,
        "Unimproved": 0
    })
)

effects["HH_head_sex"] = new_data["HH_head_sex"].map({
    "Male": 0,
    "Female": 1
})

effects["media_exposure"] = new_data["media_exposure"].map({
    "Exposed": 1,
    "Not exposed": 0
})

effects["internet_use"] = new_data["internet_use"].map({
    "Yes": 1,
    "No": 0
})

effects["residence"] = new_data["residence"].map({
    "Urban": 1,
    "Rural": 0
})

# Check all categories mapped successfully
assert effects.notna().all().all(), (
    "Some predictor categories were not recognised."
)

# =====================================================
# 3. Construct weighted simulation score
# =====================================================

# Normalise importance values across selected predictors
weights = pd.Series(feature_importance)
weights = weights / weights.sum()

# Centre predictors around their sample means
effects_centered = effects - effects.mean()

# Assumed effect strength for synthetic simulation
effect_strength = 3.0

score = (
    effects_centered[weights.index]
    .mul(weights, axis=1)
    .sum(axis=1)
    * effect_strength
)

# =====================================================
# 4. Calibrate overall ECD prevalence
# =====================================================

target_prevalence = 0.57

intercept = brentq(
    lambda b: expit(b + score).mean()
              - target_prevalence,
    -20,
    20
)

new_data["ECD_probability"] = expit(
    intercept + score
)

# =====================================================
# 5. Generate binary ECD outcome
# =====================================================

rng = np.random.default_rng(42)

new_data["ECD"] = rng.binomial(
    n=1,
    p=new_data["ECD_probability"].to_numpy()
)

# =====================================================
# 6. Validation
# =====================================================

assert new_data["ECD"].isin([0, 1]).all()

print("Overall ECD prevalence (%):")
print(round(new_data["ECD"].mean() * 100, 2))

print("\nECD prevalence by child age (%):")
print(
    new_data.groupby("child_age")["ECD"]
    .mean()
    .mul(100)
    .round(2)
)

print("\nECD prevalence by maternal education (%):")
print(
    new_data.groupby("education_level")["ECD"]
    .mean()
    .mul(100)
    .round(2)
)

# Remove intermediate probability column
new_data.drop(
    columns=["ECD_probability"],
    inplace=True
)
```

```python
# First-born children have no preceding birth interval
new_data.loc[
    new_data["birth_order"] == "1",
    "preceding_birth_interval"
] = "No preceding birth"

new_data["EaChDisease"] = np.where(
    (new_data["diarrhea"] == "Yes") |
    (new_data["cough"] == "Yes"),
    "Yes",
    "No"
)
```

```python
# =====================================================
# 9. Validation
# =====================================================

# 1. Dataset dimensions
assert len(new_data) == N

# 2. Missing values
assert new_data.isna().sum().sum() == 0

# 3. Child age categories
assert new_data["child_age"].isin(
    ["24-35", "36-47", "48-59"]
).all()

# 4. ECD outcome (binary)
assert new_data["ECD"].isin([0, 1]).all()

# 5. Birth order categories
assert new_data["birth_order"].isin(
    ["1", "2-3", "4-6", ">=7"]
).all()

# 6. First birth must have no preceding birth interval
assert (
    new_data.loc[
        new_data["birth_order"] == "1",
        "preceding_birth_interval"
    ] == "No preceding birth"
).all()

# 7. Early childhood disease consistency
# Apply only if EaChDisease is defined as cough OR diarrhea
assert (
    new_data["EaChDisease"].eq("Yes")
    ==
    (
        new_data["diarrhea"].eq("Yes") |
        new_data["cough"].eq("Yes")
    )
).all()

# 8. Display results
print("Shape:", new_data.shape)
print("Total missing:", new_data.isna().sum().sum())

print("\nECD distribution:")
print(
    new_data["ECD"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nECD prevalence by age:")
print(
    new_data.groupby("child_age")["ECD"]
    .mean()
    .mul(100)
    .round(2)
)

print("\nSample observations:")
print(new_data.head())
```
