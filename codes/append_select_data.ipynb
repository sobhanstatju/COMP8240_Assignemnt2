```python
import pandas as pd
from pandas.io.stata import StataReader
from collections import Counter

file_path = kenya_file

# Read dataset and Stata metadata
with StataReader(file_path, convert_categoricals=False) as reader:
    kenya = reader.read()
    value_labels = reader.value_labels()

    variable_to_label = dict(
        zip(reader._varlist, reader._lbllist)
    )

# Convert labelled variables into categories
for variable, label_set in variable_to_label.items():

    if variable not in kenya.columns:
        continue

    if label_set not in value_labels:
        continue

    mapping = value_labels[label_set]

    if not mapping:
        continue

    # Identify duplicate labels
    counts = Counter(mapping.values())

    # Add numeric codes only to duplicate labels
    unique_mapping = {
        code: (
            f"{label} [code {code}]"
            if counts[label] > 1
            else label
        )
        for code, label in mapping.items()
    }

    # Preserve unlabelled values instead of turning them into NaN
    original = kenya[variable]
    converted = original.map(unique_mapping)

    converted = converted.where(
        converted.notna() | original.isna(),
        original.astype("object")
    )

    kenya[variable] = pd.Categorical(converted)

print("Dataset loaded successfully!")
print("Shape:", kenya.shape)

display(kenya.head())
```

```python
# import pandas as pd

# 1. Generate sampling weights
data["wt"] = data["v005"] / 1000000


# 2. Generate country-specific PSU identifiers
data["psu"] = (
    data["country"] * 100_000 + data["v021"]
).astype("int64")

# 3. Generate unique strata identifiers
data["strata"] = (
    data.groupby(["country", "v022"], dropna=True)
        .ngroup()
        .replace(-1, pd.NA) + 1
).astype("Int64")
```

```python
# print("Number of PSUs:", data["psu"].nunique())
# print("Number of strata:", data["strata"].nunique())
print("Sum of weights:", data["wt"].sum())
```

```python
# Unweighted counts
unweighted = data.groupby("country").size()

# Weighted counts
weighted = data.groupby("country")["wt"].sum()

# Comparison table
country_table = pd.DataFrame({
    "Unweighted_N": unweighted,
    "Unweighted_%": unweighted / unweighted.sum() * 100,
    "Weighted_N": weighted,
    "Weighted_%": weighted / weighted.sum() * 100
})

country_table.loc["Total"] = [
    unweighted.sum(),
    100.0,
    weighted.sum(),
    100.0]

print(country_table.round(0))
```

```python
import numpy as np

ecd_first18 = [f"ecd{i}" for i in range(21, 39)]

for var in ecd_first18:
    data[f"{var}_score"] = (
        data[var]
        .map({
            "yes": 1,
            "no": 0,
            "don't know": 0
        })
        .fillna(0)
        .astype(int)
    )

data["ecd39_score"] = data["ecd39"].map({
    "never": 1,
    "a few times a year": 1,
    "monthly": 1,
    "weekly": 1,
    "daily": 0,
    "don't know": 0
})

data["ecd40_score"] = data["ecd40"].map({
    "not at all": 1,
    "the same or less": 1,
    "more": 0,
    "a lot more": 0,
    "don't know": 0
})

score_vars = [
    f"ecd{i}_score"
    for i in range(21, 41)
]

data["ECD_score"] = data[score_vars].sum(
    axis=1,
    min_count=len(score_vars)
)
```

```python

conditions = [
    data["b19"].between(24, 29),
    data["b19"].between(30, 35),
    data["b19"].between(36, 41),
    data["b19"].between(42, 47),
    data["b19"].between(48, 59)
]

thresholds = [
    7,
    9,
    11,
    13,
    15
]

data["ECD_threshold"] = np.select(
    conditions,
    thresholds,
    default=np.nan
)

data["ECD"] = np.where(
    data["ECD_score"].isna(),
    pd.NA,
    np.where(
        data["ECD_score"] >= data["ECD_threshold"],
        "On track",
        "Not on track"
    )
)

data["ECD"] = pd.Categorical(
    data["ECD"],
    categories=["Not on track", "On track"]
)

print(data["ECD"].value_counts(dropna=False))

print(
    data["ECD"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(1)
)
```

```python
import pandas as pd
import numpy as np

# List of variables
variables = [
    "country_name",
    "child_sex",
    "child_age",
    "birth_order",
    "preceding_birth_interval",
    "baby_postnatal_check",
    "birth_wt",
    "supplemented_VitA",
    "drug_intestinal_parasite",
    "EaChDisease",
    "cough",
    "diarrhea",
    "fever",
    "stunted",
    "wasted",
    "underwt",
    "overwt",
    "current_age",
    "education_level",
    "fathers_edu",
    "marital_status",
    "number_ANC",
    "working_status",
    "wealth_Index",
    "No_HH_Member",
    "distance_to_facility",
    "health_insurance",
    "water_source",
    "sanitation_facility",
    "HH_head_sex",
    "media_exposure",
    "internet_use",
    "residence"
]

# ECD is already coded: 1 = On track, 0 = Not on track
ecd = pd.to_numeric(data["ECD"], errors="coerce")

def make_summary_table(data, var):

    group = data.groupby(var, observed=True, sort=False)

    # Unweighted and weighted frequencies
    unweighted = group.size()
    weighted = group["wt"].sum()

    # Unweighted ECD prevalence
    ecd_unweighted = group.apply(
        lambda g: ecd.loc[g.index].mean() * 100,
        include_groups=False
    )

    # Weighted ECD prevalence
    def weighted_prevalence(g):
        y = ecd.loc[g.index]
        w = g["wt"]

        valid = y.notna() & w.notna() & (w > 0)

        if not valid.any():
            return np.nan

        return np.average(
            y[valid].astype(float),
            weights=w[valid]
        ) * 100

    ecd_weighted = group.apply(
        weighted_prevalence,
        include_groups=False
    )

    # Create table
    table = pd.DataFrame({
        "Frequency (n)": unweighted,
        "Frequency (%)": unweighted / unweighted.sum() * 100,
        "Weighted_N": weighted,
        "Weighted (%)": weighted / weighted.sum() * 100,
        "ECD Prevalence (%)": ecd_unweighted,
        "Weighted ECD Prevalence (%)": ecd_weighted
    })

    # Total row (for observations with nonmissing category)
    subset = data.loc[data[var].notna()]
    y_total = ecd.loc[subset.index]
    w_total = subset["wt"]

    valid = y_total.notna() & w_total.notna() & (w_total > 0)

    table.loc["Total"] = [
        unweighted.sum(),
        100.0,
        weighted.sum(),
        100.0,
        y_total.mean() * 100,
        np.average(
            y_total[valid].astype(float),
            weights=w_total[valid]
        ) * 100 if valid.any() else np.nan
    ]

    # Add variable and category columns
    table = table.reset_index().rename(
        columns={var: "Category"}
    )

    table.insert(0, "Variables", var)

    # Format frequencies as integers
    table["Frequency (n)"] = (
        table["Frequency (n)"].round().astype(int)
    )

    table["Weighted_N"] = (
        table["Weighted_N"].round().astype(int)
    )

    # Round percentages to 1 decimal
    pct_cols = [
        "Frequency (%)",
        "Weighted (%)",
        "ECD Prevalence (%)",
        "Weighted ECD Prevalence (%)"
    ]

    table[pct_cols] = table[pct_cols].round(1)

    return table


# Generate tables for all variables
final_table = pd.concat(
    [make_summary_table(data, var) for var in variables],
    ignore_index=True
)

print(final_table.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

# Ensure ECD is binary: 1 = On track, 0 = Not on track
data["ECD"] = pd.to_numeric(data["ECD"], errors="coerce")

def weighted_ecd_ci(g):
    valid = (
        g["ECD"].isin([0, 1])
        & g["wt"].notna()
        & (g["wt"] > 0)
        & g["psu"].notna()
    )

    d = g.loc[valid].copy()

    if d.empty:
        return pd.Series({
            "Prevalence (%)": np.nan,
            "CI Lower (%)": np.nan,
            "CI Upper (%)": np.nan
        })

    # Weighted prevalence
    p = np.average(d["ECD"], weights=d["wt"])

    # Linearised contribution of each observation
    d["linearized"] = d["wt"] * (d["ECD"] - p)

    # Sum within PSU
    psu_sum = d.groupby("psu")["linearized"].sum()

    m = len(psu_sum)

    if m < 2:
        se = np.nan
    else:
        variance = (
            m / (m - 1)
            * np.sum((psu_sum - psu_sum.mean()) ** 2)
            / d["wt"].sum() ** 2
        )
        se = np.sqrt(variance)

    z = norm.ppf(0.975)

    lower = max(0, p - z * se) if np.isfinite(se) else np.nan
    upper = min(1, p + z * se) if np.isfinite(se) else np.nan

    return pd.Series({
        "Prevalence (%)": p * 100,
        "CI Lower (%)": lower * 100,
        "CI Upper (%)": upper * 100
    })


# Calculate country-level statistics
group = data.groupby("country_name", observed=True)

unweighted = group.size()
weighted = group["wt"].sum()

ci_results = pd.DataFrame({
    country: weighted_ecd_ci(g)
    for country, g in group
}).T

table = pd.DataFrame({
    "Unweighted_N": unweighted,
    "Weighted_N": weighted,
    "ECD Prevalence (%)": ci_results["Prevalence (%)"],
    "CI Lower (%)": ci_results["CI Lower (%)"],
    "CI Upper (%)": ci_results["CI Upper (%)"]
})

# Total row
total_ci = weighted_ecd_ci(data)

table.loc["Total"] = [
    unweighted.sum(),
    weighted.sum(),
    total_ci["Prevalence (%)"],
    total_ci["CI Lower (%)"],
    total_ci["CI Upper (%)"]
]

# Format for presentation
table["Unweighted_N"] = table["Unweighted_N"].astype(int)
table["Weighted_N"] = table["Weighted_N"].round().astype(int)

table["ECD Prevalence (95% CI)"] = table.apply(
    lambda r: (
        f'{r["ECD Prevalence (%)"]:.1f} '
        f'({r["CI Lower (%)"]:.1f}, {r["CI Upper (%)"]:.1f})'
    ),
    axis=1
)

final_table = table[
    ["Unweighted_N", "Weighted_N", "ECD Prevalence (95% CI)"]
]

print(final_table)
```
